## Goal
Prototype the KisanAI Orchestrator.
It routes requests to specialized components instead of putting every function in one model.

```python
class CropAgent:
    def __init__(self, model_artifact):
        self.model = model_artifact["model"]
        self.features = model_artifact["features"]

    def recommend(self, data):
        import pandas as pd
        x = pd.DataFrame([data])[self.features]
        pred = self.model.predict(x)[0]
        result = {"crop": pred}
        if hasattr(self.model, "predict_proba"):
            result["confidence"] = float(self.model.predict_proba(x).max())
        return result
```

## Intent routing

```python
class KisanOrchestrator:
    def __init__(self, crop_agent, weather_agent, market_agent, rag_agent):
        self.crop = crop_agent
        self.weather = weather_agent
        self.market = market_agent
        self.rag = rag_agent

    def route(self, intent, **kwargs):
        if intent == "crop":
            return self.crop.recommend(kwargs["data"])
        if intent == "weather":
            return self.weather.get_weather(kwargs["location"])
        if intent == "market":
            return self.market.analyze(kwargs["crop"])
        if intent == "knowledge":
            return self.rag.answer(kwargs["question"])
        return {"error": "Unknown intent"}
```
