## Simple Weather Agent prototype

```python
class WeatherAgent:
    def __init__(self, dataframe):
        self.data = dataframe

    def summary(self, row):
        result = {}
        for key in ["temperature", "temp", "humidity", "rainfall", "rain", "wind"]:
            matches = [c for c in self.data.columns if c.lower() == key]
            if matches:
                result[matches[0]] = row[matches[0]]
        return result

agent = WeatherAgent(weather)
print("Weather agent initialized.")
```
